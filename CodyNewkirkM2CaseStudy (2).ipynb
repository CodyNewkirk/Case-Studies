```python
# File Name: StudentAwards.ipynb
# Course: SDEV220
# Author: Cody Newkirk

# This app collects student names and GPAs and checks if they qualify
# for the Dean's List or Honor Roll. Entering 'ZZZ' as a last name
# will stop the program.

while True:
    # Ask for student's last name
    last_name = input("Enter the student's last name (or 'ZZZ' to quit): ")
    if last_name.upper() == 'ZZZ':
        break

    # Ask for student's first name
    first_name = input("Enter the student's first name: ")

    # Ask for student's GPA
    try:
        gpa = float(input("Enter the student's GPA: "))
    except ValueError:
        print("Invalid input. GPA must be a number.")
        continue

    # Check GPA and print appropriate message
    if gpa >= 3.5:
        print(f"{first_name} {last_name} has made the Dean's List.")
    elif gpa >= 3.25:
        print(f"{first_name} {last_name} has made the Honor Roll.")
    else:
        print(f"{first_name} {last_name} does not qualify for the Dean's List or Honor Roll.")

    print("-" * 50)  # separator for clarity
```
